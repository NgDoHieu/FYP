## 5. Apply Random Selection, Permutation, and Shuffle

This section demonstrates three related ideas:
- selecting random elements with `choice`
- creating a random permutation with `permutation`
- shuffling an array in place with `shuffle`

They all involve randomness, but they are used for slightly different purposes in data science and ML workflows.

```python
# Random selection and permutation
arr = np.array([10, 20, 30, 40, 50])
random_pick = np.random.choice(arr, size=3, replace=False)
print("Random pick:", random_pick)

perm = np.random.permutation(arr)
print("Permutation:", perm)

shuffle_arr = arr.copy()
np.random.shuffle(shuffle_arr)
print("Shuffled in place:", shuffle_arr)

print("Original array after shuffle:", arr)
```

## 4. Generate Random Integers and Categorical Samples

`np.random.randint(low, high, size)` creates integer values in a specified range. `np.random.choice` can also generate categorical samples using probabilities via the `p` parameter.

This is useful for discrete simulation, labels, and class assignments.

```python
# Random integers from a bounded range
ints = np.random.randint(0, 10, size=(2, 5))
print("Random integers:\n", ints)

# Weighted categorical sampling
choices = np.array(['A', 'B', 'C', 'D'])
weights = np.array([0.1, 0.2, 0.3, 0.4])
category_samples = np.random.choice(choices, size=10, p=weights)
print("\nCategorical samples:", category_samples)
```

## 3. Sample from Common Random Distributions

NumPy can sample from several probability distributions:
- `uniform(low, high, size)` for evenly distributed values
- `normal(loc, scale, size)` for a bell-shaped distribution
- `standard_normal(size)` for a normal distribution with mean 0 and standard deviation 1

These distributions are useful for simulation, initialization, and understanding data generation.

```python
# Uniform distribution
u = np.random.uniform(0, 10, size=(3, 4))

# Normal distribution
n = np.random.normal(loc=0.0, scale=1.0, size=(3, 4))

# Standard normal distribution
s = np.random.standard_normal((3, 4))

print("Uniform mean/std:", u.mean(), u.std())
print("Normal mean/std:", n.mean(), n.std())
print("Standard normal mean/std:", s.mean(), s.std())

print("\nUniform sample:\n", u)
print("\nNormal sample:\n", n)
```

## 2. Generate 1D, 2D, and 3D Random Arrays

`np.random.random(shape)` creates floating-point values in the range `[0, 1)`. The shape can be 1D, 2D, or 3D, and arrays can be inspected with attributes like `.shape`, `.dtype`, `.mean()`, `.min()`, and `.max()`.

This section demonstrates how the array dimensions change the structure of the data.

```python
# 1D, 2D, and 3D random arrays
x1 = np.random.random(8)
x2 = np.random.random((3, 4))
x3 = np.random.random((2, 3, 4))

print("x1 shape:", x1.shape, "dtype:", x1.dtype, "min/max:", x1.min(), x1.max())
print("x2 shape:", x2.shape, "dtype:", x2.dtype, "mean:", x2.mean())
print("x3 shape:", x3.shape, "dtype:", x3.dtype, "std:", x3.std())

# Alternative API
rng = np.random.default_rng(7)
y1 = rng.random(8)
y2 = rng.random((2, 5))
print("\nGenerator arrays:")
print(y1)
print(y2)
```

## 1. Set Up NumPy and Reproducibility

This section shows two common ways to create reproducible random numbers:
- `np.random.seed(...)` for the legacy global random state
- `np.random.default_rng(...)` for the recommended modern generator API

Using a fixed seed makes results repeatable, which is important for teaching, debugging, and testing.

```python
import numpy as np

print("NumPy version:", np.__version__)

# Reproducibility with legacy global random state
np.random.seed(42)
print("Legacy random sample:", np.random.random(5))

# Modern generator-based randomness
rng = np.random.default_rng(42)
print("Generator sample:", rng.random(5))
```
